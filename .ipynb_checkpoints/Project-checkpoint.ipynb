# Section 1 - Background/Motivation

In traditional portfolio optimization, Mean-Variance Optimization (MVO) is often used as a benchmark for portfolio efficiency. Its reliance on single-point estimates of returns and covariances can make it sensitive to input errors; this limitation can be addressed by Monte Carlo simulation, which evaluates portfolio performance across a wide range of randomly generated scenarios. 

The end goal of both this is to create a diversified portfolio that is able to generate returns during volatile periods with the core intution being that when one stock performs poorly, the portfolio should have another asset to accomodate for that. The end goal of both approaches is to create a diversified portfolio that can generate stable returns even during volatile periods, based on the intuition that when one asset underperforms, others in the portfolio can offset the loss.

This raises the question of how a ‘diversified portfolio’ should be defined, and whether stocks from the same industry reduce diversification benefits. This project aims to use network analysis on Fortune 500 companies to explore whether incorporating network relationships into portfolio construction can lead to improved risk-adjusted outcomes.

# Section 3 - Modern Portfolio Optimization (MPT)

Modern Portfolio Theory (MPT) seeks to find the optimal combination of stocks to created a diversifed portfolio. MPT uses variance as a measure of risk for a stock; the higher the variance, the more volatile the stock is (the more the price fluctuates). It does this by calculating the covariance matrix between between stocks and looking for pairs (x, y) where COV(x, y) is close to -1 signaling that they behave oppositely. 

```python
print(returns_2023.head())
print(returns_2024.head())
```

## Section 3.1: Calculated Covariance Matrix

The covariance matrix is a matrix which relates behaviors of one stock with that of another stock. The covariance between two assets \(i\) and \(j\) is:

$$
\text{cov}(R_i, R_j) = \frac{1}{T-1} \sum_{t=1}^{T} (R_{i,t} - \bar{R}_i)(R_{j,t} - \bar{R}_j)
$$

The covariance matrix for \(n\) assets is:

$$
\Sigma =
\begin{bmatrix}
\text{cov}(R_1, R_1) & \text{cov}(R_1, R_2) & \dots & \text{cov}(R_1, R_n) \\
\text{cov}(R_2, R_1) & \text{cov}(R_2, R_2) & \dots & \text{cov}(R_2, R_n) \\
\vdots & \vdots & \ddots & \vdots \\
\text{cov}(R_n, R_1) & \text{cov}(R_n, R_2) & \dots & \text{cov}(R_n, R_n)
\end{bmatrix}
$$

Where:

- \(R_{i,t}\) = return of asset \(i\) at time \(t\)  
- \(\bar{R}_i\) = mean return of asset \(i\)  
- \(T\) = total number of periods (trading days)

For this test, the testing period is days within the fiscal year 2023 for which yfinance was able to pull close price data. We have already converted the value to a percentage return, reflective of the precentage increase (positive %) or decrease (negative %) from the previous day's closing price. The pandas library already has a .cov() function that we can call. However, we must first remove the "Date" column from the dataframe.


```python
# calculate the average returns for each stock over a year

# drop the "Date" column
returns_2023.drop(columns=['Date'], inplace=True)
print(returns_2023.head())

# calculate the matrix
covMatrix_2023 = returns_2023.cov()
print(covMatrix_2023.head())


```

## Section 3.2: Portfolio Optimization

To find the "optimal portfolio", we need to know the specific **stocks** (or assets) and each stock's **weight**. When considering *how many stocks* to have within a portfolio, it's important to consider the implications: adding more stocks reduces idiosyncratic risk, or the possibility of bad outcome for a singular stock impacing the entire portfolio. Studies have shown that most diversification benefit can be achieved with around 20-30 stocks in large-cap US portfolio. 

There is much that goes into deciding what stocks are actually contribute to the best portfolio, and this is the notion we will test later. For now, select a random sample of **10** stocks to use in our portfolio.

The next step now is to identify what the **weight** of each stock should be in the portfolio, or in other words, how much of the portfolio should each stock make up. To do so, there are many approahces. The most popular is Mean-Variance Optimization. Simply put, we are looking for the weights that will result in the portfolio having the highest *sharpe ratio*, which is also known as the tangent portfolio which lies on the *efficient frontier* where it intersects with the *Capital Allocaiton Line (CAL)* representing possible combinations of a risk-free asset. A more detailed explination of these topics can be found in this video: https://www.youtube.com/watch?time_continue=346&v=fWnyg0UeQkg&embeds_referring_euri=https%3A%2F%2Fgemini.google.com%2F&embeds_referring_origin=https%3A%2F%2Fgemini.google.com&source_ve_path=Mjg2NjY

In order to visualize the CAL and efficient frontier, we need to plot our portfolio at different test weights. The formula for Portfolio Expected Return is given by:
$$E_p = \sum_{i=1}^{n} w_i R_i$$

The formula for Portfolio Variance is given by:
$$\sigma_p^2 = \mathbf{w}^T \Sigma \mathbf{w}$$

However, risk is measured in standard deviation as follows:
$$\sigma_p = \sqrt{\mathbf{w}^T \Sigma \mathbf{w}}$$


```python
import random
from itertools import combinations_with_replacement

# random sample of 10 stocks
# samplePortfolio = random.sample(covMatrix_2023.columns.tolist(), 10) 
samplePortfolio = ['CAH', 'HD', 'THO', 'CMI', 'HUM', 'PM', 'NEM', 'CTVA', 'WSO', 'HBAN'] # this comobo produces a nice bullet
print(f'Our sample portfolio: {samplePortfolio}')

# calculate the expected return for these stocks
expectedReturn_sample = returns_2023[samplePortfolio].mean() * 250 # annualize
print(f'Expected Returns for Sample Portfolio: {expectedReturn_sample}')

# get just a list of the expected returns
expectedReturns = expectedReturn_sample.values.tolist()

# get covariance matrix of sample portfolio
samplePortfolioCOV = returns_2023[samplePortfolio].cov()
# print(samplePortfolioCOV)

# to visualize, we will iterate each possible weight in increments of 0.1 for each stock
n = 10
step = 0.1
portfolioWeights = []

def generate_weights(n, total):
    if n == 1:
        yield [total]
    else:
        for i in range(total + 1):
            for rest in generate_weights(n-1, total - i):
                yield[i] + rest

for x in generate_weights(n, 10):
    w = np.array(x) * step
    portfolioWeights.append(w)

# print(len(portfolioWeights))
# print(portfolioWeights[:10])

# calculate points 
returns = [] # y-values
risks = [] # x-values

for combo in portfolioWeights:
    comboReturn = 0
    for i in range(n):
        comboReturn += expectedReturns[i] * combo[i]
        
    returns.append(comboReturn*100)
    risks.append(np.sqrt(combo.T @ samplePortfolioCOV @ combo)*100)

    
# plot
plt.figure(figsize=(10, 5))
plt.scatter(risks, returns, color='blue', alpha = 0.3, s=10)
plt.xlabel("Risk (STD of Portfolio in %)")
plt.ylabel("Return (%)") 
plt.title("Sample Portfolio (2023)")
plt.show()
```

From here, there are two ways to calculate the most optimal portfolio. We can either draw the CAL line starting from the point of risk free return on the y-axis and see where it hits frontier, or solve for the highest Sharpe Ratio given by:
$$
\text{Sharpe Ratio} = \frac{E[R_p] - R_f}{\sigma_p}
$$

While the risk free rate is not static, we can assume 4% for the purposes of this test.

```python
# calculate sharpe ratio for every point
highestSharpe = [0, 0, 0] # [risk, return, sharpe score]
riskFree = 4

for i in range(len(returns)):
    # print(f'Progress: {i/len(returns)}%', end='\r')
    sharpe = (returns[i] - riskFree)/risks[i]
    if sharpe > highestSharpe[2]:
        highestSharpe[1] = returns[i]
        highestSharpe[0] = risks[i]
        highestSharpe[2] = sharpe

# plot
plt.figure(figsize=(10, 5))
plt.scatter(risks, returns, color='blue', alpha = 0.3, s=10)
plt.plot([0, highestSharpe[0], 2], [4, highestSharpe[1], 4+2*highestSharpe[2]], color="black", label="CAL")
plt.scatter(highestSharpe[0], highestSharpe[1], color="red", s=100, label="Optimal Portfolio", marker="*")
plt.xlabel("Risk (STD of Portfolio in %)")
plt.ylabel("Return (%)") 
plt.title("Sample Portfolio (2023)")
plt.xlim(0, None)
plt.legend()
plt.show()
```

# Section 4 - Network Building

## Section 4.1: Network Construction

In network analysis, it is important to define what **nodes** are and what relationship **edges** represent. For this analysis, we will test whether *industry* affects the performance of a portfolio. Specifically, we want to look at whether having a diversified portfolio with companies who operate in different industries compared to those who operate in the same industry will yield different results in returns. To do this, we well define nodes as companies and an edge between two nodes if there is at least one overlapping industry in which they operate in. 






```python
import networkx as nx

G = nx.Graph()

# add nodes (use only ticker symbols where yfinance info was successfully retrieved)
for ticker in returns_2023.columns:
    if ticker != "Date":
        G.add_node(ticker, market_cap = company_info.loc[company_info['Ticker'] == ticker, 'MarketCap_Updated_M'].iloc[0])

print(f'Number of Nodes in Graph: {G.number_of_nodes()}')

# add edges (if at least one industry overlaps, define as list for more efficient processing)
nodes = list(G.nodes())
for i in range(len(nodes) - 1):
    for j in range(i+1, len(nodes)):
        node1 = nodes[i]
        node2 = nodes[j]

        node1_industries = company_info.loc[company_info['Ticker'] == node1, 'Industry']
        node2_industries = company_info.loc[company_info['Ticker'] == node2, 'Industry']

        common = bool(set(node1_industries) & set(node2_industries))

        if common:
            G.add_edge(node1, node2)

print(f'Number of Edges in Graph: {G.number_of_edges()}')
```

# Section 5 - Testing

## Section 4.1: Defining Network Evaluation Criteria 

The goal of this test is to identify whether we can derive better portfolio optimization using MPT given x number of stocks using metrics derived from a network formed by Fortune 500 stocks. Given a network, we can examine each node's (stock) relative importance based on metric such as size, connectivity, and degree. In specific network analysis terms, we will define each node's eigenvector centrality, degree centrality, betweeness centrality, closeness centrality, and node size (market cap).







```python
# calculate centralities
degree_centrality = nx.degree_centrality(G)
betweenness_centrality = nx.betweenness_centrality(G)
closeness_centrality = nx.closeness_centrality(G)
eigenvector_centrality = nx.eigenvector_centrality(G)

def top_20(d):
    return {k: round(v, 3) for k, v in sorted(d.items(), key=lambda x: x[1], reverse=True)[:20]}


print("Top 20 stock scores in each centrality measure:")
print(f'Degree Centrality: \n {top_20(degree_centrality)} \n')
print(f'Betweeness Centrality: \n {top_20(betweenness_centrality)} \n')
print(f'Closeness Centrality: \n {top_20(closeness_centrality)} \n')
print(f'Eigen Vector Centrality: \n {top_20(eigenvector_centrality)} \n')


# calculate by node size
top_nodes = sorted(G.nodes(data=True), key=lambda x: x[1].get('market_cap', 0), reverse=True)[:20]
data = {node: data['market_cap'] for node, data in top_nodes}
print(f'\nTop 20 stock by market cap: {data}')

```

We will not consider betweenness centrality as an evaluation metric because all values are 1 with a heavily clustered graph like this one. 

We will consider the clustering and formation of communities in two ways: either all stocks will be part of the same community, or they will all be part different communities.

## Section 4.2: Defining Helper Functions

Here, we will define helper functions each metric individually: degree centrality, closeness centrality, eigenvector centrality, node size (market cap), and not/in same cluster community. Additionally, we will test basic structure for Monte Carlo



```python
# define function to gen random portfolio
def gen_randPortfolio(numStocks):
    samplePortfolio = random.sample(covMatrix_2023.columns.tolist(), numStocks)
    return samplePortfolio

# get historical returns for portfolio stocks
def get_PortHistReturns(portfolio):
    return returns_2023.loc[:, returns_2023.columns.isin(portfolio)]

# compute return of optimal portfolio
import cvxpy as cp
import numpy as np

def compute_optimal_weights(portStockReturns, covMatrix):
    n = len(portStockReturns)
    w = cp.Variable(n)

    # Minimize portfolio variance
    objective = cp.Minimize(cp.quad_form(w, covMatrix))

    # Constraints: weights sum to 1, no shorting
    constraints = [cp.sum(w) == 1, w >= 0]

    # Solve the problem
    prob = cp.Problem(objective, constraints)
    prob.solve()

    weights = w.value
    port_return = np.dot(weights, portStockReturns)
    port_vol = np.sqrt(weights.T @ covMatrix @ weights)
    

    return weights, port_return, port_vol


# define function to return average of given metric across the portfolio

def compute_AvgMetric(portfolio, metric):
    sum_returns = 0
    
    if metric == "degree":
        for stock in portfolio:
            sum_returns += degree_centrality[stock]

        return sum_returns/len(portfolio)

    if metric == "closeness":
        for stock in portfolio:
            sum_returns += closeness_centrality[stock]
    
        return sum_returns/len(portfolio)

    if metric == "eigenvector":
        for stock in portfolio:
            sum_returns += eigenvector_centrality[stock]
    
        return sum_returns/len(portfolio)

    if metric == "marketcap":
        for stock in portfolio:
            sum_returns += G.nodes[stock].get("market_cap")
    
        return sum_returns/len(portfolio)

    return
        
    
```

```python
# test monte carlo structure
numIterations = 1
numStocks = 10

for i in range(numIterations):
    # generate portfolio
    portfolio = gen_randPortfolio(numStocks)

    portHistReturns = get_PortHistReturns(portfolio)
    expected_returns = portHistReturns.mean(axis=0).values

    cov_matrix_subset = covMatrix_2023.loc[portfolio, portfolio].values

    weights, val1, val2 = compute_optimal_weights(expected_returns, cov_matrix_subset)
    
    print(
        np.round(weights, 4),
        round(val1, 6),
        round(val2, 4)
    )

    for metric in ["degree", "closeness", "eigenvector", "marketcap"]:
        print(compute_AvgMetric(portfolio, metric))
    
    
```

## Section 4.3: Monte-Carlo Simulation

Here, we will formalize our testing for Monte Carlo Simulation. On a high level, Monte Carlo Simulation is using randomness to estimate uncertainty or probabilities. Our uncertainty here is whether or not network statistics may play some use in portfolio construction. To test this, we will run the simulation multiple times on random portfolios and record their return and average given network statistic (degree, closeness, eigenvalue, etc). 

```python
# set runtime parameters

num_iterations = 5000
portfolio_size = 10 # number of stocks in portfolio

portfolios_returns = []
portfolios_degree = []
portfolios_closeness = []
portfolios_eigenvector = []
portfolios_marketcap = []
portfolios_numConnections = []


# formal testing method
def test_MonteCarlo(num_iterations, portfolio_size):
    
    for i in range(num_iterations):
        portfolio = gen_randPortfolio(numStocks)

        portHistReturns = get_PortHistReturns(portfolio)
        expected_returns = portHistReturns.mean(axis=0).values
    
        cov_matrix_subset = covMatrix_2023.loc[portfolio, portfolio].values
    
        weights, val1, val2 = compute_optimal_weights(expected_returns, cov_matrix_subset)

        portfolios_returns.append(val1)
        
        for metric in ["degree", "closeness", "eigenvector", "marketcap"]:
            if metric == "degree":
                portfolios_degree.append(compute_AvgMetric(portfolio, metric))
            if metric == "closeness":
                portfolios_closeness.append(compute_AvgMetric(portfolio, metric))
            if metric == "eigenvector":
                portfolios_eigenvector.append(compute_AvgMetric(portfolio, metric))
            if metric == "marketcap":
                portfolios_marketcap.append(compute_AvgMetric(portfolio, metric))


        # count num communities
        portfolios_numConnections.append(G.subgraph(portfolio).number_of_edges())


    # optional print
    # print([float(round(x, 4)) for x in portfolios_returns])
    # print([float(round(x, 4)) for x in portfolios_degree])
    # print([float(round(x, 4)) for x in portfolios_closeness])
    # print([float(round(x, 4)) for x in portfolios_eigenvector])
    # print([float(round(x, 4)) for x in portfolios_marketcap])
    # print([float(round(x, 4)) for x in portfolios_communities])

    print("Done")



test_MonteCarlo(num_iterations, portfolio_size)
```
